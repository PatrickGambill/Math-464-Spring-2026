## Numbers ##

In Python we can work with numbers. There are a few types of numbers in Python. For our purposes, we will use ints and floats.

**Ints:** These are integers. We use them for indices and the like. They have arbitrary precision.

**Floats:** These are floating point numbers. (Specifically double point precision floating point numbers). These have limited precision. For most practical purposes, these are fine but if you are working with numbers that vary a lot in range or need extreme precision, you may need other numbers. Occasionally they create rounding errors with matrix oeprations too.

```python
#We are creating a variable in python by typing, name = ...

#By default any number without a decimal is an int in Python. mynum is an int
mynum = 12332
#By default any number with a decimal is a float in Python. myfloat is a float
myfloat = 1.2

#We can print a string to output text
print("Variable Values:")
#We can print a variable to output its value. The f"..." syntax is a shortcut.
#You can also call print(mynum) if you want mynum without the text in front.
print(f"mynum: {mynum}")
print(f"myfloat: {myfloat}")
#This print statement adds a line of space in the output
print("\n")

#If you are ever unsure what type a variable is, call type to see what type of variable we are working with
print("Variable Types:")
print(type(mynum))
print(type(myfloat))
print("\n")

#We can do standard arithmatic with numbers. Python defaults to converting these to floats if we mix types
print("Arithmetic:")
print(f"Addition: {mynum + myfloat}")
print(f"Subtraction: {mynum - myfloat}")
print(f"Multiplication: {mynum * myfloat}")
print(f"Division: {mynum / myfloat}")

#Exponentiation has different notation than you might expect
print(f"Exponentiation: {mynum ** myfloat}")
print("\n")

#Modular Arithmetic and Floor Division
print("Modular Arithmetic and Floor Division (Compare this to division):")
#This is the remainder if we take mynum / myfloat
print(f"Modular Arithmetic: {mynum % myfloat}")
#This is the whole number part of mynum / myfloat
print(f"Floor Division: {mynum // myfloat}")
print("\n")

#If we wish to convert between types, we can call int and float
print("Type Conversion")
#Notice that myfloat gets rounded down to the nearest int
print(int(myfloat))
print(float(mynum))
print(type(int(myfloat)))
print(type(float(mynum)))
```

## Lists ##

Lists are a datatype built on "lower level" data types. These store an ordered collection of objects. These objects need not be the same.

```python
#Here is a list
mylist = [12,432,7834,21.3434,'my string',[1,21,4]]

#We can get elements of lists
#This returns the first object (lists always start at index 0 in Python)
print("First Object")
print(mylist[0])
print("\n")

#This returns the last object
print("Last Object")
print(mylist[-1])
print("\n")

#This allows us to replace the object in the 3rd index of our list with 4.16
mylist[3] = 4.16
print("New List")
print(mylist)
print("\n")

#We can concatenate lists. This takes one list and attaches it to the end of another
list1 = [1,2,3,4,5]
list2 = ['a','b','c','d','e','f']

print("Concatenated")
print(list1+list2)
print("\n")

#We can also get a slice of a list. This returns part of a list.
#Slices always include the object in the first index and exclude the objects in the last index

print("Slicing")
print(list2[1:3])

#Notice the object in index -1 is not included
print(list2[2:-1])

#If we want to incluide the whole tail end of a list we do this:
print(list2[2:])
#If we want to include the whole begining of a list, we do list:
print(list2[:4])
```

## For Loops ##

For loops allow for repeated execution of code. These help save a lot of time.

```python
#We don't want to have to type repeated operations by hand. for loops help here.
#In Python, the spacing matters for "for" loops. Any indentation indicates the code goes inside the loop.

#Lists are allowed to be empty
mylist = []
#This tells Python to run the loop 8 times. It starts with i=0 and ends with i=7.
for i in range(8):

  #Append adds an object to the end of a list.
  mylist.append(3*i+7)

print(mylist)
print("\n")

#You can iterate over any iterable. These include lists, dictionaries, sets, ect.
print("The objects in the mylist:")
for i in mylist:
  print(i)
print("\n")

#Suppose we wish to add all of the objects in a long list
longlist = [1,2,6,1,7,123,8,12312] * 5
print("The long list:")
print(longlist)
print("\n")

#We can use x to store our sum
x = 0
for i in longlist:

  #This updates x to be the (old value of x) + (i)
  x = x+i

print(f"The sum of all of the entries of longlist is: {x}")

#We also have the sum function. This sums all objects in a list.
tot = sum(longlist)
print(f"The sum of all of the entries of longlist is: {tot}")
```

## Conditional Statements ##

Conditional Statements run code if a given condition is met. These are if, elif, and else statements

**if:** An if statement runs the given code if the condition is met

**elif:** An elif statement always follows after an if or another elif statment. This gives a second criteria to check, if the above statements are not satisfied

**else:** An else statement always follows all of the if and elif statements. This will run if none of the if or elif statements were satisfied above.

```python
#Consider this list
mylist = [123,6,2323,787878,124567,2,1,-6445,123]

#We want to make a list of numbers in mylist that are greater than 1000
greaterthan = []
lessthan = []

for i in mylist:
  #This is an if statement
  if i > 1000:
    greaterthan.append(i)
  #This else statement runs when the if statement isn't satisfied
  else:
    lessthan.append(i)

print(greaterthan)
print(lessthan)
```

```python
#Consider this list
mylist = [123,6,2323,787878,124567,2,1,-6445,123,14,16,24,-2345123]

#We want to break it into equivalence classes based on the remainder mod4
mod0 = []
mod1 = []
mod2 = []
mod3 = []

for i in mylist:

  if i%4 == 0:
    mod0.append(i)
  #We can have many elif statements
  elif i%4 == 1:
    mod1.append(i)
  elif i%4 == 2:
    mod2.append(i)
  else:
    mod3.append(i)

print(f"Remainder 0: {mod0}")
print(f"Remainder 1: {mod1}")
print(f"Remainder 2: {mod2}")
print(f"Remainder 3: {mod3}")
```

## Numpy ##

Numpy is a commonly used library in Python. It allows for us to do lots of math operations efficiently. It is (probably) the most commonly used library in Python for linear algebra.

```python
#You may have to run this to install numpy. For most big libraries, this allows you to install the library
!pip install numpy
```

```python
#This imports numpy. To call functions from numpy, you use the syntax, np.function(...)

import numpy as np
#This creates a numpy array from a list. Numpy arrays function similar to matrices.
#They must contain the same data type throughout
np.array([12,2])

#In an array, all objects must be the same type. Numpy will automatically convert types where possible.
x = np.array([12,12.4,3])
y = np.array([6543,7,5])

print("x and y")
print(f"x: {x}")
print(f"y: {y}")
print("\n")

#Numpy applies all array operations elementwise.
print("Arithmetic")
print(f"x+y: {x+y}")
print(f"x-y: {x-y}")
print(f"x*y: {x*y}")
print(f"x/y: {x/y}")
print("\n")

#Getting objects and slicing still applies similar to lists
print("Getting Objects from List")
print(x[1])
print(x[1:3])
print("\n")

#We can index from an array as well. This will only return the objects at the given indices
print("Indexing")
myarray = np.array([123,6,23,7,34232,7878,231461,0,0,1])
indexarray = np.array([4,6,9])

print(myarray[indexarray])
```

```python
#Numpy has a great deal of matrix operations. To work with these, we create a 2D array by making a list of lists

print("Matrix Basics")
print("\n")
print("The Matrices")
mat1 = np.array([[1,2,3],[4,5,6],[7,8,90]])
print(f"mat1: {mat1}")

#We can get the dimensions of the array
print(f"Shape: {mat1.shape}")

mat2 = np.array([[5,6,7],[4,6,1],[87,2,5],[0,1,2]])
print(f"mat2: {mat2}")
print(f"Shape: {mat2.shape}")
print("\n")

#We need to be careful to use @ when multiplying matrices. Otherwise we get elementwise operations
print("Matrix Multiplication")
print(f"mat2 X mat1: {mat2@mat1}")
print("\n")

#This looks like x^TAy
print(f"x^T X mat1 X y: {x@mat1@y}")
print("\n")

#Two different ways to compute dot products
print("Dot Products of x and y")
print(x@y)
print(np.dot(x,y))
print("\n")

#We can find the inverse too
print("Inverse")
inv = np.linalg.inv(mat1)
print(f"mat1's inverse: \n{mat1@inv}")
#Note: For some matrices, the inverse may have numerical errors.
#These depend on the conditioning of the matrix as well as the numbers involved (compared to the floating point precision)
#Be careful in general. We have classes at WSU to cover these topics on occasion

#We get entries near 0!
#We can round to fix this
rounded = np.round(mat1@inv,8)
print(f"Rounded inverse: \n{rounded}")
```

```python
#We can also get individual rows/columns of an array
print("mat1")
print(mat1)
print("\n")

print("Getting Rows and Columns")
#Get the row at index 1
print(f"Row 1: {mat1[1]}")
print("\n")
#get the column at index 1
print(f"Column 1: {mat1[:,1]}")
print("\n")

print("Submatrix (Top Left Corner Only)")
#We can get sub matrices
print(mat1[0:2,0:2])
```

```python
#We can do more complicated slicing. Consider the matrix created below:
A = []

for i in range(20):
  newrow = []
  for j in range(45):

    newrow.append(i*j+7)

  A.append(newrow)

#This is A. It is quite large and printing it will not tell us much
A = np.array(A)

#These are the indices we want from A
indices = np.array([0,2,5,5,8,12,8,9,11,12,24,15,12,0,12,2,3])

#We are returning the columns from A with the indices given from indices
B = A[:,indices]
print(B)
```
