## An BIP Formulation ##

Consider the knapsack problem we formulated in lecture. Suppose we have a collection of gem values and weights given below:

```python
import pulp

#Given Information
Gems = [str(i) for i in range(1, 40)]
Values = {i: 32*(int(i)%6) + 3 for i in Gems}
Weights = {i: 14*(int(i)%11) + 5 for i in Gems}
MaxWeight = 200

#Create the problem
prob = pulp.LpProblem("Knapsack", pulp.LpMaximize)

decision_vars = pulp.LpVariable.dicts("Items", Gems, 0, 1, pulp.LpInteger)

#Objective
prob += (pulp.lpSum([Values[i]*decision_vars[i] for i in Gems]), "Total_Value")

#Constraint
prob += (pulp.lpSum([Weights[i]*decision_vars[i] for i in Gems]) <= MaxWeight,"Weight_Constraint")

prob.writeLP("Knapsack.lp")
prob.solve()

print("Status:", pulp.LpStatus[prob.status])

print("These are the gems selected")
for v in prob.variables():
  #Only output the values for non-zero variables
  if v.varValue > 0:
    print(v.name, "=", v.varValue)
```

```python
#Let's show the total value
print("The total value is:")
print(pulp.value(prob.objective))
print("The total weight is:")
print(pulp.lpSum([Weights[i]*decision_vars[i].varValue for i in Gems]))

#If you print the value of the constraint, you get the left side - the right side
print(pulp.value(prob.constraints["Weight_Constraint"]))
```
