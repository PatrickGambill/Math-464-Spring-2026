# Pandas #

When reading in datasets, Pandas is a helpful library in Python. Run the cell below first to import pandas.

```python
import pandas as pd
```

Below, we are going to do a few common operations on made up data. None of the people listed are current of former students of mine. I am making up fake data about a few family members, fictional characters, and famous people in order to avoid FERPA violations.

```python
#Read in the CSV. Make sure the path is a string.
#If you are in Google collab, be sure to mount drive first
#(The folder icon on the left side of the screen is where you can mount drive)

#Replace this string with your path inside the read_csv function
df=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Math 464 Spring 2026/Lectures/Attendence Sheet - Sheet1.csv')

#Get the first few rows of the data frame
#(For large datasets, you want the first few rows rather than all of them)
print(df.head())

#With missing data or bad data, you need to be sure to fill in the missing data with the appropriate null entry

#Fill all missing grades with F grades
df['Grade'] = df['Grade'].fillna('F')
#Fill in missing Passing information with 0
df['Passing'] = df['Passing'].fillna(0)
#Fill in missing last name information with unknown
df['Last Name'] = df['Last Name'].fillna('Unknown')

#Here, we can convert the datatype of the passing column into bools.
#A float is not an appropriate datatype for a yes/no question
df['Passing'] = df['Passing'].astype(bool)

#See how all the data is cleaned now
print(df.head())

#Get a list from a column. .tolist() is faster and goes all the way down.
#This is the preferred method to convert from numpy arrays/pandas dataframes into lists.
lastnames = df['Last Name'].tolist()
print(lastnames)
```
