## Numpy Practice and Tricks ##

We will spend today working with Numpy arrays for practice.

```python
import numpy as np
```

```python
#Suppose we have this array
A = np.array([[-i+j for i in range(8)] for j in range(11,3,-1)])

print(A)

#We can find the indices of the negative entries
negatives = np.where(A < 0)
print(negatives)

#And we can find the corresponding values by taking A[negatives]
print(A[negatives])
```

```python
#Suppose we have a 1D array
B = np.array([1,-4,3,-6,12,-1,5])

#These are the indices corresponding to negative entries of B.
#We need the [0] at the end since we have a 1D array rather than a 2D array
negativeindices = np.where(B<0)[0]
print(negativeindices)
#This will give the maximum negative value of B
print(np.max(B[negativeindices]))

#Notice we can call B[negativeindices] to only return the negative entries of B
print(B[negativeindices])
#If we want the corresponding index of B, we first get the index of the returned array
smallestindexhelper = np.argmax(B[negativeindices])
#We get the corresponding index of the original array by calling this index on negativeindices
print(negativeindices[smallestindexhelper])
```

```python
#We can create arrays of a specific type of data by specifying dtype as an argument
A = np.zeros((4,6),dtype=np.bool)
B = np.ones((6,6),dtype=np.bool)
C = np.ones((4,4),dtype=np.int32)

#We can stack one array on top of another
AB = np.vstack((A,B))
print(AB)

#We can stack two arrays side by side. Notice the type is converted to the most general one
CA = np.hstack((C,A))
print(CA)
```

1) Consider the Fibonacci numbers: $F_0 = 0, F_1 = 1, F_n = F_{n-1} + F_{n-2}$ for all $n \geq 2$.  Find the index of the first Fibonacci number greater than $(10)^{5}$ using numpy arrays and functions (i.e. no for loops). (You can use to for loops to get the array of Fibonacci numbers).

2) Compute the first $100$ non-zero Fibonacci numbers $A$. Compute the first $15$ factorials (starting with $1!$) $B$. Stack them in an array $[A B]$ and multiply each entry of the array by its index. Compute the total sum of entries in the resulting array.

```python
#Get the first 100 non-zero Fibonacci numbers for use in the remaining parts
Fibs = [1,1]
while len(Fibs) < 100:
  Fibs.append(Fibs[-1] + Fibs[-2])

A = np.array(Fibs)

#1
indices = np.where(A > 10**5)[0]
print(f'The first Fibonacci number > 10^5 is the {indices[0]} Fibonacci number')

#2
#Get our B array
import math
B = np.array([math.factorial(i) for i in range(1,16)])

#Stack A and B
AB = np.hstack((A,B))
#Get the array of indices
index = np.array([i for i in range(len(AB))])

multiplied = AB*index
print(f'The sum is: {sum(multiplied)}')

print(f'The sum is: {np.dot(AB,index)}')
```
